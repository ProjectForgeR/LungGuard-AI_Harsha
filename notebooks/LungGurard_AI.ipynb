```python
import os
import shutil
from sklearn.model_selection import train_test_split

source_base = "/content/drive/MyDrive/LungGuard_AI/Dataset/IQ-OTH-NCCD/Processed_Images/The IQ-OTHNCCD lung cancer dataset"

output_base = "/content/drive/MyDrive/LungGuard_AI/Dataset/IQ-OTH-NCCD/Split_Data"

class_mapping = {
    "Bengin cases": "Benign",
    "Malignant cases": "Malignant",
    "Normal cases": "Normal"
}

for original_class, clean_class in class_mapping.items():

    source_folder = os.path.join(source_base, original_class)

    images = [
        f for f in os.listdir(source_folder)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    train_files, temp_files = train_test_split(
        images,
        test_size=0.30,
        random_state=42,
        shuffle=True
    )

    val_files, test_files = train_test_split(
        temp_files,
        test_size=0.50,
        random_state=42,
        shuffle=True
    )

    splits = {
        "train": train_files,
        "validation": val_files,
        "test": test_files
    }

    for split_name, file_list in splits.items():

        destination = os.path.join(
            output_base, split_name, clean_class
        )

        os.makedirs(destination, exist_ok=True)

        for filename in file_list:
            shutil.copy2(
                os.path.join(source_folder, filename),
                os.path.join(destination, filename)
            )

    print(f"\nClass: {clean_class}")
    print("Training:", len(train_files))
    print("Validation:", len(val_files))
    print("Testing:", len(test_files))

print("\nDataset splitting completed!")
```

```python
import tensorflow as tf

DATA_DIR = "/content/drive/MyDrive/LungGuard_AI/Dataset/IQ-OTH-NCCD/Split_Data"

IMG_SIZE = (224, 224)
BATCH_SIZE = 8

train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR + "/train",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=True,
    seed=42
)

val_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR + "/validation",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR + "/test",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="categorical",
    shuffle=False
)

print("\nClass names:", train_ds.class_names)
print("Training dataset loaded successfully!")
print("Validation dataset loaded successfully!")
print("Testing dataset loaded successfully!")
```

```python
import tensorflow as tf

IMG_SIZE = (224, 224)

base_model = tf.keras.applications.EfficientNetB0(
    weights="imagenet",
    include_top=False,
    input_shape=(224, 224, 3)
)

base_model.trainable = False

print("EfficientNetB0 loaded successfully!")
print("Pretrained weights: ImageNet")
print("Base model trainable:", base_model.trainable)
print("Input shape:", base_model.input_shape)
```

```python
from tensorflow.keras import layers, models

# Input layer
inputs = layers.Input(shape=(224, 224, 3))

# EfficientNetB0 feature extraction
x = base_model(inputs, training=False)

# Classification layers
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)

# Three output classes
outputs = layers.Dense(
    3,
    activation="softmax",
    name="predictions"
)(x)

# Build complete model
model = models.Model(
    inputs=inputs,
    outputs=outputs,
    name="LungGuard_AI_EfficientNetB0"
)

# Compile model
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
import numpy as np
from sklearn.utils.class_weight import compute_class_weight

class_names = ["Benign", "Malignant", "Normal"]

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=np.array([0, 1, 2]),
    y=np.array(
        [0] * 84 +
        [1] * 392 +
        [2] * 291
    )
)

class_weight_dict = {
    i: float(weight)
    for i, weight in enumerate(class_weights)
}

print("Class Weights:")
for i, name in enumerate(class_names):
    print(name, ":", class_weight_dict[i])
```

```python
import os
import tensorflow as tf

MODEL_DIR = "/content/drive/MyDrive/LungGuard_AI/Models"
os.makedirs(MODEL_DIR, exist_ok=True)

callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=3,
        restore_best_weights=True,
        verbose=1
    ),

    tf.keras.callbacks.ModelCheckpoint(
        filepath=os.path.join(MODEL_DIR, "lungguard_best.keras"),
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.2,
        patience=2,
        min_lr=0.00001,
        verbose=1
    )
]

print("Training callbacks configured successfully!")
print("Early stopping: Enabled")
print("Best model checkpoint: Enabled")
print("Learning rate reduction: Enabled")
```

```python
EPOCHS = 5

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    class_weight=class_weight_dict,
    callbacks=callbacks,
    verbose=1
)

print("\nLungGuard AI training completed!")
```

```python
import tensorflow as tf
import os

best_model_path = "/content/drive/MyDrive/LungGuard_AI/Models/lungguard_best.keras"

# Load best saved model
best_model = tf.keras.models.load_model(best_model_path)

# Evaluate on untouched test dataset
test_loss, test_accuracy = best_model.evaluate(
    test_ds,
    verbose=1
)

print("\n========== LUNGGUARD AI TEST RESULTS ==========")
print("Test Accuracy:", round(test_accuracy * 100, 2), "%")
print("Test Loss:", round(test_loss, 4))
print("Total Test Images:", 166)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    classification_report,
    confusion_matrix
)

class_names = ["Benign", "Malignant", "Normal"]

# Generate predictions
y_prob = best_model.predict(test_ds)
y_pred = np.argmax(y_prob, axis=1)

# Get actual labels
y_true = np.concatenate([
    labels.numpy().argmax(axis=1)
    for images, labels in test_ds
])

# Classification report
print("========== CLASSIFICATION REPORT ==========")

print(classification_report(
    y_true,
    y_pred,
    labels=[0, 1, 2],
    target_names=class_names,
    zero_division=0,
    digits=4
))

# Confusion matrix
cm = confusion_matrix(
    y_true,
    y_pred,
    labels=[0, 1, 2]
)

print("========== CONFUSION MATRIX ==========")
print(cm)

# Plot confusion matrix
plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("LungGuard AI - Confusion Matrix")
plt.tight_layout()
plt.show()
```
